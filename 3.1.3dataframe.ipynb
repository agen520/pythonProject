## dataframe创建、属性、访问、方法

### 通过series创建

```python
import pandas as pd
import numpy as np
s1=pd.Series([1,2,3,4,5,6])
s2=pd.Series([7,8,9,10,11,12])
df=pd.DataFrame({'第一列':s1,'第二列':s2})
df
```

### 通过字典创建

```python
df=pd.DataFrame({'id':[1,2,3,4,5,],'name':['clq','pll','dandi','huang','hou'],'age':[18,19,20,21,22],'score':[150,149,148,147,149]})
df
```
