#ndarray学习

##1.ndarray特性

```python
import numpy as np
```

```python
arr=np.array(5)
print(arr)
print("维度是：",arr.ndim)
print("形状是：",arr.shape)
```

```python
arr=np.array([1,2,3,4])
print(arr)
print("维度是：",arr.ndim)
```

```python
arr=np.array([[1,2,3],[2,3,3]])
print(arr)
print("维度是：",arr.ndim)
```

#同质性

```python
arr=np.array([1,"hello"])#强制转换为相同的性质
print(arr)
```
