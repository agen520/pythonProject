# pandas分为series（一维）和dataframe（二维）

## series创建、属性、访问、方法

### 创建：通过列表创建和通过字典创建
通过列表创建时，index要包在series的括号里,,,,,,
通过字典创建时，记得字典要加大括号

```python
#通过列表创建
import pandas as pd
s=pd.Series([1,2,3,4,5,6],index=['a','b','c','d','e',"f"],name="clqnb666")
print(s)
```

```python
#通过字典创建
s=pd.Series({"a":1,"b":2,"c":3,"d":4,"e":5,"f":6},name="clqnb666")
print(s)
```

```python
#通过已有字典创建
s1=pd.Series(s,index=['a','b','c'])
print(s1)
```

## 属性index、values、dtype、shape、ndim、size、name、loc[]、iloc[]、at[]、iat[]

## loc[]、iloc[]、at【】、iat【】的用法:没加i的是根据标签索引,【】里要加“”。加i的是根据位置索引,[]不需要加“”（这里的位置指的第一个位置的下标为0,1,2,3,4这种）

```python
print(s.index)
print(s.values)
print(s.dtype)
print(s.shape)
print(s.ndim)
print(s.size)
print(s.loc["a"])
print(s.iloc[0])
print(s.at["b"])
print(s.iat[0])
```

# 访问数据：布尔索引，head（），tail（）

```python
# print(s.head(3))
# print(s.tail(4))
print(s[s<3])
```
