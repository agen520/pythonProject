# 案例1、创建一个包含10名学生的series，成绩范围在50-100之间。计算平均分、最高分、最低分，并找出高于平均分的人数

### 创建学生列表的方法：注意：str后面跟的符号是（），一开始写错成【】了

```python
import numpy as np
import pandas as pd
np.random.seed(40)
clq=np.random.randint(50,101,10)
s=pd.Series(clq,index=['学生'+str(i)for i in range(1,11)])
print(s)
pll=s.mean()
print('平均分为：',s.mean())
print('最高分为：',s.max())
print('最低分为：',s.min())
print('高于平均分的人数为：',len(s[s>pll]))
```

# 案例2：温度系统分析：找出温度超过30度的天数，计算平均温度，将温度从高到低排序，找出温度变化最大的两天

```python
temperture=pd.Series([28,31,29,32,30,27,33],index=['1','2','3','4','5','6','7'],name='温度表')
print('温度超过30的天数为：',len(temperture[temperture>30]))
print('平均气温为：',temperture.mean())
print('将温度从高到低的排序的结果是：',temperture.sort_values(ascending=False))
temperture.diff().abs().sort_values(ascending=False).keys()[:2].tolist()
```
